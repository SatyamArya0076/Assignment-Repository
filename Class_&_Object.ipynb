```python
class Student:
  def addDetails(self, name, age, course): #Local Variable
    # Use 'self' to bind attributes to the current object
    self.n = name   # Instance variable in python
    self.a = age
    self.c = course

  def showDetails(self):
    # Use 'self' to access attributes of the current object
    print(self.n)
    print(self.a)
    print(self.c)

object1 = Student()
object1.addDetails("Amit", 23, "Python")
object1.showDetails()

print("="*20)

object2 = Student()
object2.addDetails("Priya", 22, "java")
object2.showDetails()
```
